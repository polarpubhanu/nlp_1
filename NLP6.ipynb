```python
import nltk

# Tokenizer
nltk.download('punkt')
nltk.download('punkt_tab')

# POS Tagger
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')

# NER Chunker (NEW REQUIRED)
nltk.download('maxent_ne_chunker')
nltk.download('maxent_ne_chunker_tab')

# Other datasets
nltk.download('words')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')



```

```python
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag, ne_chunk
import string
```

```python
text = "Apple is looking at buying a UK startup for $1 billion. John works at Microsoft."
print("Original Text:\n", text)
```

```python
# Sentence Tokenization
sent_tokens = sent_tokenize(text)
print("Sentence Tokens:\n", sent_tokens)

# Word Tokenization
word_tokens = word_tokenize(text)
print("\nWord Tokens:\n", word_tokens)
```

```python
# Lowercase
lower_text = text.lower()

# Remove punctuation
normalized_text = lower_text.translate(str.maketrans('', '', string.punctuation))

print("Normalized Text:\n", normalized_text)
```

```python
stop_words = set(stopwords.words('english'))

filtered_words = [word for word in word_tokens if word.lower() not in stop_words]

print("After Stop-word Removal:\n", filtered_words)
```

```python
pos_tags = pos_tag(word_tokens)

print("POS Tags:\n", pos_tags)
```

```python
lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(word) for word in word_tokens]

print("Lemmatized Words:\n", lemmatized_words)
```

```python
named_entities = ne_chunk(pos_tags)

print("Named Entities:\n")
print(named_entities)
```
